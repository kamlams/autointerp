```python
import pickle
# token_ls structure: token_ls[batch_id][prompt_id][token_position]
token_ls_path = "token_list_260904.pkl"
with open(token_ls_path, "rb") as f:
    token_ls = pickle.load(f)

print("num_batches:", len(token_ls))
print("num_prompts_per_batch:", len(token_ls[0]))
print("num_tokens_per_prompt:", len(token_ls[0][0]))
```

```python
import torch
# feature_info structure: [batch_id, prompt_id, token_position, activation]
feature_info = torch.load("feature28617__to__260904.pt")
print(feature_info.shape)

top10_acts, top10_indices = torch.topk(feature_info[:,-1], dim=-1, k=10)
for i in top10_indices:
    batch_id, prompt_id, token_pos, activation = feature_info[i]
    batch_id = batch_id.int()
    prompt_id = prompt_id.int()
    token_pos = token_pos.int()

    cur = token_ls[batch_id][prompt_id][token_pos]
    print(f"activation:{activation} token:\033[31m{cur}\033[0m")
    print(f"last 15 tokens: {token_ls[batch_id][prompt_id][max(token_pos - 15, 0) : token_pos]}")
    print(f"\033[36m{''.join(token_ls[batch_id][prompt_id][max(token_pos - 15, 0) : max(token_pos - 1, 0)])}\033[31m{cur}\033[0m")
```
